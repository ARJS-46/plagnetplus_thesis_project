# Figure 6 — frozen W05 final test evaluation



```python
from pathlib import Path
import sys

ROOT = next(path for path in [Path.cwd(), *Path.cwd().parents] if (path / "src" / "plagnetplus").is_dir())
sys.path.insert(0, str(ROOT / "src"))
OUTPUT = ROOT / "figures" / "final"
OUTPUT.mkdir(parents=True, exist_ok=True)

import pandas as pd
from IPython.display import display
from plagnetplus.figures.w05_test import plot_canopy_biplot, plot_cluster_composition

REFERENCE = ROOT / "results/reference"
metrics = pd.read_csv(REFERENCE / "w05_test/w05_final_test_metrics.csv")
display(metrics.T.rename(columns={0: "W05 test"}))

from IPython.display import Image as IPyImage, Markdown

def show_saved_figure(path):
    path = Path(path)
    display(Markdown(f"**Saved figure:** `{path}`"))
    display(IPyImage(filename=str(path)))
```

```python
scores = pd.read_csv(REFERENCE / "figure_06/w05_train_validation_test_canopy_pca_scores.csv")
loadings = pd.read_csv(REFERENCE / "figure_06/w05_train_validation_test_canopy_pca_loadings.csv")
figure = plot_canopy_biplot(scores, loadings)
out_path = OUTPUT / "figure_06a_w05_test_canopy_biplot.png"
figure.savefig(out_path, dpi=300)
show_saved_figure(out_path)
```

```python
assignments = pd.read_csv(REFERENCE / "figure_06/w05_train_validation_test_cluster_assignments.csv")
figure = plot_cluster_composition(assignments)
out_path = OUTPUT / "figure_06b_w05_test_cluster_composition.png"
figure.savefig(out_path, dpi=300)
show_saved_figure(out_path)
```
