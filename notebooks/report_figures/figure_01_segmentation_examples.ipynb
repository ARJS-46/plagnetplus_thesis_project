```python
from pathlib import Path
import sys

ROOT = next(path for path in [Path.cwd(), *Path.cwd().parents] if (path / "src" / "plagnetplus").is_dir())
sys.path.insert(0, str(ROOT / "src"))
OUTPUT = ROOT / "figures" / "final"
OUTPUT.mkdir(parents=True, exist_ok=True)

import os
from plagnetplus.figures.segmentation import plot_segmentation_example

DATA_ROOT = Path(os.environ["PNP_DATA_ROOT"])
RAW_ROOT = Path(os.environ["PNP_RAW_SCAN_ROOT"])
EXAMPLES = {
    "methana": (
        RAW_ROOT / "Methana/Met91-78-9-XII/Met91-78-9-XII_00001_00002.JPG",
        DATA_ROOT / "interim/plagnetplus/non_aegina_clean_labels_v1/14__Methana__Met91-78-9-XII/Methana/Met91-78-9-XII_00001_00002/clean_instance_labels_uint16.png",
    ),
    "fct": (
        RAW_ROOT / "FCT/6861_FCT/6861_00001_00002.JPG",
        DATA_ROOT / "interim/plagnetplus/fct_6861_figure_mask_v1/clean_instance_labels_uint16.png",
    ),
}

from IPython.display import Image as IPyImage, Markdown, display

def show_saved_figure(path):
    path = Path(path)
    display(Markdown(f"**Saved figure:** `{path}`"))
    display(IPyImage(filename=str(path)))
```

```python
for name, (scan, mask) in EXAMPLES.items():
    figure = plot_segmentation_example(scan, mask)
    out_path = OUTPUT / f"figure_01_{name}_segmentation.png"
    figure.savefig(out_path, dpi=300)
    show_saved_figure(out_path)
```
